```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
#loading the entire dataset
imdb_dataset = pd.read_csv("IMDB-Movie-Data.csv");
```

```python
top_rating_movies = imdb_dataset.nlargest(6, 'Rating')[['Title', 'Rating']]
plt.figure(figsize=(16,5))
plt.title("Top 6 Movies Having Rating")
plt.barh(top_rating_movies["Title"],top_rating_movies["Rating"])
plt.ylabel("Movie Name")
plt.xlabel("Rating")
plt.grid()
plt.show()
```

```python
top_revenue_movies = imdb_dataset.nlargest(6, 'Revenue (Millions)')[['Title', 'Revenue (Millions)']]
plt.figure(figsize=(16,5))
plt.title("Top 6 Movies Having Revenue (Millions)")
plt.plot(list(top_revenue_movies['Revenue (Millions)']),list(top_revenue_movies['Title']),marker='o', linestyle='-', color='b')
plt.show()
```
