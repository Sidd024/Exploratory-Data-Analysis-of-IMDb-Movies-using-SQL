```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
#loading the entire dataset
imdb_dataset = pd.read_csv("IMDB-Movie-Data.csv");
```

```python
#Top 6 Movies Title & Rating
top_rating_movies = imdb_dataset.nlargest(6, 'Rating')[['Title', 'Rating']]
plt.figure(figsize=(16,5))
plt.title("Top 6 Movies Having Rating")
plt.barh(top_rating_movies["Title"],top_rating_movies["Rating"])
plt.ylabel("Movie Name")
plt.xlabel("Rating")
plt.grid()
plt.show()
```

```python
#Top 6 Movies Title & Revenue (Millions)
top_revenue_movies = imdb_dataset.nlargest(6, 'Revenue (Millions)')[['Title', 'Revenue (Millions)']]
plt.figure(figsize=(16,5))
plt.title("Top 6 Movies Having Revenue (Millions)")
plt.plot(list(top_revenue_movies['Revenue (Millions)']),list(top_revenue_movies['Title']),marker='o', linestyle='-', color='b')
plt.show()
```

```python
#Top 6 Movies Genre & Count
genre_counts = (
    imdb_dataset['Genre']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)
plt.figure(figsize=(10,5))
plt.barh(genre_counts.index, genre_counts.values)
plt.xlabel("Number of Movies")
plt.ylabel("Genre")
plt.title("Top 10 Movie Genres on IMDb")
plt.gca().invert_yaxis()
plt.grid(axis='x')
plt.show()
```
